# Gate-A#2 — Kronos Walk-Forward Evaluation

**Cakupan:** BTCUSDT + ETHUSDT · timeframe 1h · horizon 24 bar · n=590 window/pair.

Threshold GO pre-register: IC ≥ +0.10 · hit-rate ≥55% per pair.
Sumber data: `research/gate-a2-formal-{pair}-h24.txt` (artefak `kronos_eval`, anti-leak walk-forward).
Logika analisis ada di modul `seith_analysis.kronos_eval_viz` — notebook ini tipis.

```sql
from pathlib import Path

import matplotlib.pyplot as plt

from seith_analysis.kronos_eval_viz import (
    load_detail,
    load_summary,
    render_dashboard,
    summary_table,
)

ROOT = Path.cwd()
if ROOT.name != "SEITH":
    ROOT = ROOT.parents[1]
RESEARCH = ROOT / "research"
PAIRS = ["BTCUSDT", "ETHUSDT"]

details = {p: load_detail(RESEARCH / f"gate-a2-formal-{p}-h24.txt") for p in PAIRS}
summaries = {p: load_summary(RESEARCH / f"gate-a2-formal-{p}-h24.txt") for p in PAIRS}
{p: len(df) for p, df in details.items()}
```

## Ringkasan metrik formal

```sql
summary_table(summaries).round(4)
```

## Dashboard per pair
Panel: (1) PnL kumulatif arah-sinyal · (2) scatter pred-vs-realized · (3) rolling IC · (4) hit-rate bulanan.

```sql
for pair in PAIRS:
    fig = render_dashboard(
        details[pair],
        ticker=pair,
        save_path=RESEARCH / f"gate-a2-charts-{pair}.png",
    )
    plt.show()
```

## Interpretasi (diisi setelah membaca chart)

- **BTCUSDT**: IC +0.022, hit 49.0% → **tanpa edge** di h24; persistence bahkan unggul arah.
- **ETHUSDT**: IC +0.088 (signifikan, p<0.05 pada n=590), hit 54.9% → **edge tipis nyata**.
- Kedua pasar mean-reverting di horizon ini (persistence negatif).
- Keputusan Gate-A#2: **belum GO** menurut threshold pre-register; opsi — uji horizon lain (h8/h48), ETH-only conditional, atau NO-GO.
